###Simulate environmental perturbations

```python
def add_confusing_object(env,task,obj_id, obj_type):
  __, __, obj_size = env.info[obj_id]
  if obj_type=="block":
    obj_template = 'box/box-template.urdf'
    params={'DIM': obj_size}
  elif obj_type=="stand":
    obj_template = 'stacking/stand.urdf'
    params={'DIM': obj_size}
  else:
    obj_size = task.get_random_size(0.05, 0.3, 0.05, 0.3, 0.05, 0.05)
    #zone_pose = task.get_random_pose(env, zone_size)
    obj_template = 'container/container-template.urdf'
    half = np.float32(obj_size) / 2
    params = {'DIM': obj_size, 'HALF': half}
  
  urdf = task.fill_template(obj_template, params)
  addition_pose = task.get_random_pose(env, obj_size)
  addition_id=env.add_object(urdf, addition_pose)
  icolor = p.getVisualShapeData(obj_id)[0][-1]
  p.changeVisualShape(addition_id, -1, rgbaColor=icolor)
  os.remove(urdf)
  return addition_id
```
